```python
# Author: Alberto Fernández Comesaña

import os
import matplotlib.pyplot as plt

import utils
from sklearn.model_selection import cross_val_score
```

```python
DISCHARGES_DIR = os.path.join(os.pardir, 'discharges')
METEO_DIR = os.path.join(os.pardir, 'meteo-csv')

X_train, Y_train, X_test, Y_test = utils.load_data(METEO_DIR, DISCHARGES_DIR, 'B')
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
cv_score = cross_val_score(lr, X_train, Y_train, cv=10, scoring='r2').mean()
print(f'Mean cross validation: {cv_score:.4f}')

lr.fit(X_train, Y_train)
Y_pred = lr.predict(X_test)

utils.plot_predictions(Y_test, Y_pred)
```
